# VISUALISATION

```python
# HISTOGRAM
sns.histplot(data=df,x="age",kde=True)
plt.show()
```

# Observation: 
The histogram shows the distribution of passenger ages. The average age is approximately 29.7 years, and the median age is 28 years. The ages range from 0.42 to 80 years among the recorded values. Since 177 age values are missing, the age distribution represents only passengers with recorded ages.

# Finding: 
The dataset includes passengers from infants to elderly adults, with the central age values around the late twenties.

```python
#distribution of numerical columns
df.hist(figsize=(15, 10), bins=20, edgecolor="black")
plt.suptitle("Distribution of Numerical Columns")
plt.show()
```

# Observation: 
The numerical distributions show differences in passenger age, fare, family size, and passenger class. The fare column has a mean of approximately 32.20 and a median of approximately 14.45, indicating a right-skewed distribution. Most passengers paid relatively low fares, while a smaller number paid much higher fares.
# Finding:
Passenger fares vary considerably, and the mean fare is influenced by high-fare observations.

```python
sns.histplot(data=df,x="fare",kde=True)
plt.show()
```

# Observation: 
The fare histogram shows that many passengers paid lower fares, while fewer passengers paid very high fares. The average fare was 32.20, compared with a median of 14.45. The maximum fare was approximately 512.33.
# Finding: 
The fare distribution is strongly right-skewed, suggesting that a small number of passengers paid substantially more than most others.

```python
#BAR PLOT
plt.figure(figsize=(7,5))
ax=sns.barplot(data=df,x="sex",y="survived")
ax.set_title("Survival Rate by Gender")
ax.set_xlabel("Gender")
ax.set_ylabel("Survival Rate")
ax.set_ylim(0,1)
plt.tight_layout()
plt.show()
```

# Observation: 
The survival rate was approximately 74.2% for female passengers and 18.9% for male passengers.
# Finding: 
Female passengers had a substantially higher survival rate than male passengers in this dataset. Gender was strongly associated with survival.

```python
plt.figure(figsize=(7,5))
ax=sns.barplot(data=df,x="pclass",y="survived",order=[1,2,3])
ax.set_title("Survival Rate by Passenger Class")
ax.set_xlabel("Passenger Class(1=First,2=Second,3=Third)")
ax.set_ylabel("Survival Rate")
plt.show()
```

# Observation:
The survival rates were approximately: First class: 63.0%,Second class: 47.3%,Third class: 24.2%
# Finding: 
Survival rates decreased from first class to third class. Passengers in higher classes had better observed survival outcomes.

```python
#CATEGORICAL COLUMNS
categorical_cols=df.select_dtypes(include=["object","category","bool"]).columns

for col in categorical_cols:
    plt.figure(figsize=(7,5))
    df[col].value_counts().head(15).plot(kind="bar",edgecolor="black")
    plt.title("count of categories")
    plt.xlabel("column")
    plt.ylabel("Count")
    plt.show()
    
```

# Observation: 
The frequency counts show that the dataset contains 577 male passengers and 314 female passengers. Third-class passengers form the largest passenger-class group, with 491 passengers. Southampton was the most common embarkation location, accounting for 644 passengers with recorded embarkation ports.
# Finding: 
The dataset contains more male than female passengers, more third-class passengers than passengers in either other class, and a predominance of passengers embarking at Southampton.

```python
# GENDER AND PASSENGER CLASS TOGETHER plt.figure(figsize=(9, 5))

ax = sns.barplot(
    data=df,
    x="pclass",
    y="survived",
    hue="sex",
    order=[1, 2, 3],
    errorbar=None
)

ax.set_title("Survival Rate by Gender and Passenger Class", fontsize=15)
ax.set_xlabel("Passenger Class")
ax.set_ylabel("Survival Rate")
ax.legend(title="Gender")
plt.tight_layout()
plt.show()
```

# Observation:
The combined bar chart reveals differences in survival rates across both gender and passenger clas
# Finding: 
Female passengers had higher survival rates than male passengers in every class. Female first-class passengers had the highest observed survival rate, while male third-class passengers had the lowest among these six groups.


```python
#BOXPLOT
sns.boxplot(data=df,x="survived",y="age")
plt.show()
```

# Observation: 
The box plot compares the age distributions of passengers who survived and those who did not survive. It displays the median age, spread of ages, and potential outliers for both groups.
# Finding: 
The plot helps assess whether passenger age distributions differed by survival status. The age box plot alone does not establish that age caused a passenger to survive or not survive.


```python
# IDENTIFY OUTLIERS
numeric_cols = df.select_dtypes(
    include=np.number
).columns

for col in numeric_cols:
    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[col])

    plt.title(f"Box Plot of {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

# Observation: 
The box plots highlight extreme observations in numerical columns such as age, fare, sibsp, and parch. The fare column is particularly likely to show high-value outliers because its maximum value is much larger than its median.
# Finding:
The numerical data contain values that differ substantially from the central distributions. These observations should be investigated before deciding whether they represent errors or valid passenger records.

```python
#SCATTER PLOT
sns.scatterplot(data=df,x="age",y="fare",hue="survived",palette={0:"red",1:"blue"})
plt.show()
```

# Observation: 
The scatter plot compares passenger age and fare, using different colors to represent survival status. It helps visualize whether survival outcomes appear to differ across age and fare combinations.
# Finding: 
The plot can reveal clusters and overlapping patterns, but age and fare alone cannot fully explain survival. The observed relationship should be interpreted alongside gender and passenger class.

```python
#RELATIONSHIP BETWEEN TWO COLUMNS
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df,x="sibsp", y="parch")

plt.title("Relationship Between Column1 and Column2")
plt.xlabel("Sibsp")
plt.ylabel("Parch")
plt.tight_layout()
plt.show()
```

# Observation: 
The scatter plot compares sibsp (siblings or spouses aboard) with parch (parents or children aboard). Many observations occur at low values, indicating that numerous passengers travelled with few or no relatives in these categories.
# Finding:
Family-group sizes vary across passengers. The plot can reveal passengers travelling alone or with relatives, although these two columns do not capture every possible family relationship.

```python
#PAIRPLOt
sns.pairplot(df[["survived","pclass","age","sibsp","fare"]].dropna(),hue="survived")
plt.show()
```

# Observation: 
The pair plot compares survived, pclass, age, sibsp, and fare. It combines individual distributions with scatter plots of pairs of variables, making it easier to inspect several relationships together.
# Finding:
Passenger class and fare are useful variables to investigate alongside survival. The pair plot also helps identify overlapping distributions and unusual observations that may require further analysis.

```python
#HEATMAP
numeric_df=df.select_dtypes(include="number")
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f")
plt.show
```

# Observation: 
The heatmap displays the strength and direction of linear correlations between numerical variables. Values closer to +1 indicate stronger positive correlation, values closer to −1 indicate stronger negative correlation, and values around zero indicate weaker linear correlation.
# Finding: 
The heatmap should be used to identify candidate relationships for further analysis. Correlation does not establish causation, and the binary survival variable should be interpreted as an indicator rather than a continuous measurement.
